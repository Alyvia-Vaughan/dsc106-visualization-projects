```python
# Since there is a good amount of outliers in the numeric columns, I am going to drop all of the rows that are above 1 standard deviation from the 75% percentile in one of the numeric columns. 
# This is because I want to be able to analyze the data without the outliers skewing the results.
numeric_cols = ["price", "price percal", "package_weight", "Protein", "Total Fat", "Carbohydrate", "Sugars, total", "Fiber, total dietary", "Sodium", "Cholesterol"]
new_data = data
for col in numeric_cols:
    upper_bound = new_data[col].quantile(0.75) + new_data[col].std()
    new_data = new_data[new_data[col] <= upper_bound]
```

```python
# Get top 3 categories per store as proportions
store_totals = new_data.groupby("store").size()
top3_per_store = (
    new_data.groupby(["store", "category"])
    .size()
    .reset_index(name="count")
)

# Divide each store's category count by that store's total
top3_per_store["proportion"] = (
    top3_per_store["count"] / top3_per_store["store"].map(store_totals)
)

# Now get top 3 per store by proportion
top3_per_store = (
    top3_per_store.groupby("store")
    .apply(lambda x: x.nlargest(3, "proportion"))
    .reset_index(drop=True)
)

# Sort so stores are clustered
top3_per_store = top3_per_store.sort_values(["store", "proportion"], ascending=[True, False])

# Create combined x label
top3_per_store["label"] = top3_per_store["store"] + "\n" + top3_per_store["category"]

# Color map per store
colors = top3_per_store["store"].map({"Target": "steelblue", "Walmart": "orange", "WholeFoods": "green"})

plt.figure(figsize=(14, 6))
plt.bar(top3_per_store["label"], top3_per_store["proportion"], color=colors)
plt.title("Top 3 Categories per Store (Proportion of Store's Total Products)")
plt.xlabel("")
plt.ylabel("Proportion")
plt.xticks(rotation=45, ha='right')

from matplotlib.patches import Patch
legend_elements = [Patch(facecolor="steelblue", label="Target"),
                   Patch(facecolor="orange", label="Walmart"),
                   Patch(facecolor="green", label="WholeFoods")]
plt.legend(handles=legend_elements, title="Store")
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.patheffects import withStroke

clean = data.dropna(subset=["FPro_class", "price"])

stores = ["Walmart", "WholeFoods"]
colors = {"Walmart": "#1f77b4", "WholeFoods": "#2ca02c"}

fig, ax = plt.subplots(figsize=(12, 6))

medians = {}
for store in stores:
    subset = clean[clean["store"] == store]
    grouped = subset.groupby("FPro_class")["price"].quantile([0.25, 0.5, 0.75]).unstack()
    medians[store] = grouped[0.50]
    
    x = grouped.index
    lower = grouped[0.25]
    upper = grouped[0.75]
    
    ax.fill_between(x, lower, upper, alpha=0.17, color=colors[store])
    ax.plot(x, grouped[0.50], color=colors[store], linewidth=2.5, marker="o",
            markersize=7, label=store)

gaps = {}
for fpro_class in [0, 1, 2, 3]:
    wf = medians["WholeFoods"][fpro_class]
    wm = medians["Walmart"][fpro_class]
    gaps[fpro_class] = wf - wm

largest_gap_class = max(gaps, key=lambda k: gaps[k])

for fpro_class in [0, 1, 2, 3]:
    wf = medians["WholeFoods"][fpro_class]
    wm = medians["Walmart"][fpro_class]
    gap = gaps[fpro_class]
    is_largest = fpro_class == largest_gap_class

    ax.annotate("", xy=(fpro_class, wf), xytext=(fpro_class, wm),
            arrowprops=dict(arrowstyle='<->',
                            color='#D32F2F' if is_largest else '#222222',
                            lw=2.5 if is_largest else 1.7))


    if is_largest:
        ax.text(fpro_class + 0.05, (wf + wm) / 2,
                f"${gap:.2f} ← largest gap",
                fontsize=12, color='#C0392B', va='center', fontweight='bold',
                path_effects=[withStroke(linewidth=0.05, foreground='#222222')])
    else:
        ax.text(fpro_class + 0.05, (wf + wm) / 2,
                f"${gap:.2f}",
                fontsize=11, color='#222222', va='center', fontweight='semibold')

# compute overall median range for each store
walmart_range = medians["Walmart"].max() - medians["Walmart"].min()
wf_range = medians["WholeFoods"].max() - medians["WholeFoods"].min()

ax.text(2.1, medians["Walmart"].mean() - 0.7,
        f"Walmart median shifts only ${walmart_range:.2f}\nacross all processing levels",
        fontsize=11, color="#1f77b4", fontweight='bold')

ax.text(0.1, medians["WholeFoods"].max() + 0.3,
        f"WholeFoods median shifts ${wf_range:.2f}\nacross all processing levels",
        fontsize=11, color="#2ca02c", fontweight='bold')

ax.set_xlabel("Processing Level (NOVA Classification)", fontsize=11, fontweight='medium')
ax.set_ylabel("Price ($)", fontsize=11, fontweight='medium')
ax.set_xticks([0, 1, 2, 3])
ax.set_xticklabels(["0\nUnprocessed or\nMinimally Processed", "1\nProcessed Culinary\nIngredients",
                    "2\nProcessed Foods", "3\nUltra-processed\nFoods"])

ax.legend(title="point = median price,  shaded region = IQR",
          fontsize=9, title_fontsize=9)

fig.suptitle("Prices Vary More By Store Than By Processing Level",
             fontsize=15, fontweight='bold', y=1.02)
fig.text(0.5, 0.95,
         "Median price and interquartile range (middle 50%) shown per store | Median and IQR are robust to extreme price outliers",
         ha='center', fontsize=12, color='#111111', fontweight='medium')

plt.tight_layout()
plt.savefig("final_viz.png", dpi=150, bbox_inches='tight')
plt.show()
```
